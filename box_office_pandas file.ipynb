```python
language_map = {
    "en": "English",
    "ja": "Japanese",
    "ko": "Korean",
    "zh": "Chinese",
    "cn": "Chinese",
    "hi": "Hindi",
    "fr": "French",
    "de": "German",
    "it": "Italian",
    "es": "Spanish",
    "sv": "Swedish",
    "da": "Danish",
    "tr": "Turkish",
    "cs": "Czech",
    "el": "Greek",
    "et": "Estonian",
    "fa": "Persian",
    "th": "Thai",
    "sr": "Serbian",
    "pt": "Portuguese",

    "ar": "Arabic",
    "fi": "Finnish",
    "id": "Indonesian",
    "kn": "Kannada",
    "ml": "Malayalam",
    "nl": "Dutch",
    "no": "Norwegian",
    "pa": "Punjabi",
    "pl": "Polish",
    "ru": "Russian",
    "ta": "Tamil",
    "te": "Telugu",
    "tl": "Filipino",
    "uk": "Ukrainian",
    "vi": "Vietnamese"
}
```

```python
df["Original_Language"] = df["Original_Language"].replace(language_map)
```

```python
df["Original_Language"].value_counts()
```

```python
df.isnull().sum()
```

```python
df["Genres"] = df["Genres"].fillna("Unknown")
```

```python

df["Rating"] = df["Rating"].astype(str)
df["Rating"] = df["Rating"].str.replace("/10", "", regex=False)
df["Rating"] = df["Rating"].astype(float)
```

```python
df["Vote_Count"] = df["Vote_Count"].fillna(0)
```

```python
df["Original_Language"] = df["Original_Language"].fillna("Unknown")
```

```python
df["Production_Countries"] = df["Production_Countries"].fillna("Unknown")
```

```python
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
```

```python
df["Rating"] = df["Rating"].fillna(df["Rating"].median())
```
